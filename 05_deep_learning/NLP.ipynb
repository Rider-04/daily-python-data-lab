```python
from sklearn.datasets import load_iris
import pandas as pd
```

```python
iris = load_iris()
```

```python
X = iris.data
```

```python
y = iris.target
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sd = StandardScaler()
```

```python
Xs=sd.fit_transform(X)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
Xreduced = pca.fit_transform(Xs)
Xr = pd.DataFrame(Xreduced)
```
