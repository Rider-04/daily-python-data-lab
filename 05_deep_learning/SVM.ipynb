```python
import numpy as np

Sugar = [0.1, 0.2, -0.3]
diabetes = [0.8, 0.21, -0.34]

n = 0
a = 0
b = 0

for i, j in zip(Sugar, diabetes):
    n = n + (i * j)
    a = a + (i * i)
    b = b + (j * j)

o = n / (np.sqrt(a) * np.sqrt(b))

print("Cosine Similarity:-", o)
```
