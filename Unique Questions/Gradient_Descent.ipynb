```python
import random
import math

def gradient_descent(x,y,w=random.randint(-100, 100),b=random.randint(-100, 100),lr=0.1):
    try:
        if len(x) == len(y):
            previous_loss = float("inf")
            tolerance = 0.001

            for epoch in range(1000):
                y_pred = []
                n = len(x)
                lf = 0
                dw = 0
                db = 0
                
                for i in range(n):
                    predict = w * x[i] + b
                    y_pred.append(predict)

                for i in range(n):
                    error = y[i] - y_pred[i]
                    lf += (error ** 2)

                for i in range(n):

                    error = y[i] - y_pred[i]
                    dw += -2 * x[i] * error
                    db += -2 * error
                    
                w = w - lr * dw

                b = b - lr * db

                
                if abs(previous_loss - lf) < tolerance:

                    print("Convergence reached")

                    break
                previous_loss = lf
            print(f"Epoch : {epoch + 1}")
            print(f"Weight : {w}")
            print(f"Bias : {b}")
            print(f"Loss : {lf}")
            print(f"Predictions : {y_pred}")
        else:
            print("Length of x and y must be same")
    except:
        print("Invalid input")
```
