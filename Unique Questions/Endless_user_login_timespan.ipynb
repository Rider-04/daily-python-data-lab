```python
from datetime import datetime
import pandas as pd


# Users
users = {
    "U1": {"username": "Parth"},
    "U2": {"username": "Rahul"}
}
logged_in_users = {}
session_data = []
while True:
    print("\n1. Enter User")
    print("2. Print Data")
    print("3. Exit")
    choice = input("\nEnter your choice (1/2/3): ")
    if choice == "1":
        user_id = input("\nEnter User ID: ")
        if user_id in users:
            username = users[user_id]["username"]
            if user_id in logged_in_users:
                login_time = logged_in_users[user_id]
                logout_time = datetime.now()
                duration = logout_time - login_time

                duration_seconds = duration.total_seconds()

                duration_minutes = duration_seconds / 60
                print("\nLogout successful!")

                print("User:", username)
                print("User ID:", user_id)
                print("Login time:", login_time)
                print("Logout time:", logout_time)
                print("Time spent:", round(duration_seconds, 2), "seconds")
                print("Time spent:", round(duration_minutes, 2), "minutes")

                session_data.append({
                    "user_id": user_id,
                    "username": username,
                    "login_time": login_time,
                    "logout_time": logout_time,
                    "duration_seconds": round(duration_seconds, 2),
                    "duration_minutes": round(duration_minutes, 2)
                })

                del logged_in_users[user_id]

            else:

                login_time = datetime.now()

                logged_in_users[user_id] = login_time

                print("\nLogin successful!")

                print("User:", username)
                print("User ID:", user_id)
                print("Login time:", login_time)


        # -----------------------------------
        # USER ID DOES NOT EXIST
        # -----------------------------------

        else:

            print("\nUser ID does not exist.")

            create_user = input("Do you want to create a new user? (yes/no): ")


            if create_user.lower() == "yes":

                username = input("Enter your username: ")


                # Find next User ID
                user_numbers = []

                for uid in users:
                    user_numbers.append(int(uid[1:]))


                if len(user_numbers) == 0:
                    new_number = 1

                else:
                    new_number = max(user_numbers) + 1


                user_id = "U" + str(new_number)


                # Create user
                users[user_id] = {
                    "username": username
                }


                print("\nRegistration successful!")
                print("Your User ID is:", user_id)


                # Automatically login new user
                login_time = datetime.now()

                logged_in_users[user_id] = login_time


                print("\nLogin successful!")
                print("User:", username)
                print("User ID:", user_id)
                print("Login time:", login_time)


    # -----------------------------------
    # 2. PRINT DATA
    # -----------------------------------

    elif choice == "2":

        # Create DataFrame
        df = pd.DataFrame(session_data)


        print("\n========== DATAFRAME ==========")

        if df.empty:

            print("No session data available.")

        else:

            print(df)


            # Save complete DataFrame to CSV
            df.to_csv("user_sessions.csv", index=False)


            print("\n========== CSV DATA ==========")

            # Read CSV again
            csv_data = pd.read_csv("user_sessions.csv")

            print(csv_data)

            print("\nData saved to user_sessions.csv")


    # -----------------------------------
    # 3. EXIT
    # -----------------------------------

    elif choice == "3":
        print("\nProgram ended.")
        break
    else:
        print("\nInvalid choice. Please enter 1, 2 or 3.")
```
