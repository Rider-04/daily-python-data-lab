```python
import math
def regression_matrices(actual,predict):
    try:
        if len(actual)==len(predict):
            n=len(actual)
            mae_total=0
            mse_total=0
            r2_total=0
            mean = math.mean(actual)
            for i in range (n):
                error=actual[i]-predict[i]
                r2_md=actual[i]-mean
                mae_total+=abs(error)
                mse_total+=(error**2)
                r2_total+=(r2_md**2)
            mae=mae_total/n
            mse=mse_total/n
            rmse=(math.sqrt(mse_total))/n
            r2_score=1-(mse_total/r2_total)
            print(f"Mean Absolute Error :- {mae}")
            print(f"Mean Squared Error :- {mse}")
            print(f"Root Mean Squared Error :- {rmse}")
            print(f"R2-Score :- {r2_score}")
        else:
            print("The size of both lists/arrays is not same")
    except:
        print("Invalid Input")
```
