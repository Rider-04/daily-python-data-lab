```python
users = {"U1": {"username": "Parth","password": "1234"},
         "U2": {"username": "Rahul","password": "5678"}}
```

```python
from datetime import datetime
import pandas as pd

print("\n1. Existing User")
print("2. New User")

choice = input("\nEnter your choice (1/2): ")

if choice == "1":
    user_id = input("\nEnter User ID: ")
    password = input("Enter Password: ")

    if user_id in users:
        if users[user_id]["password"] == password:
            username = users[user_id]["username"]
            
            print("\nLogin successful!")
            login_time = datetime.now()

            print("Login time:", login_time)

            input("\nPress ENTER when you want to logout...")
            logout_time = datetime.now()

            duration = logout_time - login_time

            duration_seconds = duration.total_seconds()
            duration_minutes = duration_seconds / 60

            print("\nLogout successful!")

            print("User:", username)
            print("User ID:", user_id)
            print("Login time:", login_time)
            print("Logout time:", logout_time)

            print("Time spent:",round(duration_seconds, 2),"seconds")
            print("Time spent:",round(duration_minutes, 2),"minutes")

            session_data = pd.DataFrame([{
                "user_id": user_id,
                "username": username,
                "login_time": login_time,
                "logout_time": logout_time,
                "duration_seconds": round(duration_seconds, 2),
                "duration_minutes": round(duration_minutes, 2)}])
            old_data = pd.read_csv("user_sessions.csv")
            final_data = pd.concat([old_data, session_data],ignore_index=True)
            final_data.to_csv("user_sessions.csv",index=False)
            print("\nSession saved to user_sessions.csv")

        else:
            print("\nIncorrect password.")
    else:
        print("\nUser ID does not exist.")

elif choice == "2":
    username = input("Enter your username: ")
    password = input("Create your password: ")
    user_numbers = []
    
    if len(user_numbers) == 0:
        user_id = "U1"

    else:
        user_id = "U" + str(len(user_numbers) + 1)


    users[user_id] = {"username": username,"password": password}


    print("\nRegistration successful!")
    print("Your User ID is:", user_id)

    print("\nLogin successful!")

    login_time = datetime.now()

    print("Login time:", login_time)

    input("\nPress ENTER when you want to logout...")

    logout_time = datetime.now()

    duration = logout_time - login_time

    duration_seconds = duration.total_seconds()
    duration_minutes = duration_seconds / 60

    print("\nLogout successful!")

    print("User:", username)
    print("User ID:", user_id)
    print("Login time:", login_time)
    print("Logout time:", logout_time)
    print("Time spent:",round(duration_seconds, 2),"seconds")
    print("Time spent:",round(duration_minutes, 2),"minutes")
    
    session_data = pd.DataFrame([{
        "user_id": user_id,
        "username": username,
        "login_time": login_time,
        "logout_time": logout_time,
        "duration_seconds": round(duration_seconds, 2),
        "duration_minutes": round(duration_minutes, 2)}])
    old_data = pd.read_csv("user_sessions.csv")
    final_data = pd.concat(
            [old_data, session_data],
            ignore_index=True)
    final_data.to_csv(
            "user_sessions.csv",
            index=False)
    print("\nSession saved to user_sessions.csv")
else:
    print("\nInvalid choice. Please enter 1 or 2.")
```

```python
from datetime import datetime
import pandas as pd

active_users = {}

while True:
    print("1. Existing User")
    print("2. New User")
    print("3. Logout User")
    print("4. View CSV")

    choice = input("\nEnter your choice (1/2/3/4): ")

    if choice == "1":
        user_id = input("\nEnter User ID: ")

        if user_id in users:
            password = input("Enter Password: ")

            if users[user_id]["password"] == password:
                username = users[user_id]["username"]

                if user_id in active_users:
                    print("\nUser is already logged in.")
                else:
                    print("\nLogin successful!")
                    login_time = datetime.now()

                    print("Login time:", login_time)

                    active_users[user_id] = {
                        "username": username,
                        "login_time": login_time
                    }

                    print("\nUser is now active.")
            else:
                print("\nIncorrect password.")
        else:
            print("\nUser ID does not exist.")
            print("\nRedirecting to New User registration...")

            username = input("\nEnter your username: ")
            password = input("Create your password: ")

            user_numbers = []

            for uid in users:
                user_numbers.append(int(uid[1:]))

            if len(user_numbers) == 0:
                user_id = "U1"
            else:
                user_id = "U" + str(max(user_numbers) + 1)

            users[user_id] = {
                "username": username,
                "password": password
            }

            print("\nRegistration successful!")
            print("Your User ID is:", user_id)

            print("\nLogin successful!")

            login_time = datetime.now()

            print("Login time:", login_time)

            active_users[user_id] = {
                "username": username,
                "login_time": login_time
            }

            print("\nUser is now active.")

    elif choice == "2":
        username = input("Enter your username: ")
        password = input("Create your password: ")

        user_numbers = []

        for uid in users:
            user_numbers.append(int(uid[1:]))

        if len(user_numbers) == 0:
            user_id = "U1"
        else:
            user_id = "U" + str(max(user_numbers) + 1)

        users[user_id] = {
            "username": username,
            "password": password
        }

        print("\nRegistration successful!")
        print("Your User ID is:", user_id)

        print("\nLogin successful!")

        login_time = datetime.now()

        print("Login time:", login_time)

        active_users[user_id] = {
            "username": username,
            "login_time": login_time
        }

        print("\nUser is now active.")

    elif choice == "3":
        user_id = input("\nEnter User ID to logout: ")

        if user_id in active_users:
            username = active_users[user_id]["username"]
            login_time = active_users[user_id]["login_time"]

            logout_time = datetime.now()

            duration = logout_time - login_time

            duration_seconds = duration.total_seconds()
            duration_minutes = duration_seconds / 60

            print("\nLogout successful!")

            print("User:", username)
            print("User ID:", user_id)
            print("Login time:", login_time)
            print("Logout time:", logout_time)

            print("Time spent:", round(duration_seconds, 2), "seconds")
            print("Time spent:", round(duration_minutes, 2), "minutes")

            session_data = pd.DataFrame([{
                "user_id": user_id,
                "username": username,
                "login_time": login_time,
                "logout_time": logout_time,
                "duration_seconds": round(duration_seconds, 2),
                "duration_minutes": round(duration_minutes, 2)
            }])

            old_data = pd.read_csv("user_sessions.csv")

            final_data = pd.concat(
                [old_data, session_data],
                ignore_index=True
            )

            final_data.to_csv(
                "user_sessions.csv",
                index=False
            )

            print("\nSession saved to user_sessions.csv")

            del active_users[user_id]

        else:
            print("\nUser is not currently logged in.")

    elif choice == "4":
        try:
            final_data = pd.read_csv("user_sessions.csv")

            print("\nUser Sessions CSV:")
            print(final_data)

        except FileNotFoundError:
            print("\nuser_sessions.csv does not exist.")

    else:
        print("\nInvalid choice. Please enter 1, 2, 3 or 4.")
```

```python
from datetime import datetime
import pandas as pd

active_users = {}

while True:
    print("\n1. Existing User")
    print("2. New User")
    print("3. Logout User")
    print("4. View CSV")

    choice = input("\nEnter your choice (1/2/3/4): ")

    # =========================
    # 1. EXISTING USER
    # =========================

    if choice == "1":
        user_id = input("\nEnter User ID: ")

        if user_id in users:
            password = input("Enter Password: ")

            if users[user_id]["password"] == password:
                username = users[user_id]["username"]

                if user_id in active_users:
                    print("\nUser is already logged in.")

                else:
                    login_time = datetime.now()

                    active_users[user_id] = {
                        "username": username,
                        "login_time": login_time
                    }

                    print("\nLogin successful!")

            else:
                print("\nIncorrect password.")

        else:
            print("\nUser ID does not exist.")
            print("\nRedirecting to New User registration...")

            username = input("\nEnter your username: ")
            password = input("Create your password: ")

            user_numbers = []

            for uid in users:
                user_numbers.append(int(uid[1:]))

            if len(user_numbers) == 0:
                user_id = "U1"
            else:
                user_id = "U" + str(max(user_numbers) + 1)

            users[user_id] = {
                "username": username,
                "password": password
            }

            login_time = datetime.now()

            active_users[user_id] = {
                "username": username,
                "login_time": login_time
            }

            print("\nRegistration successful!")
            print("Your User ID is:", user_id)
            print("Login successful!")


    # =========================
    # 2. NEW USER
    # =========================

    elif choice == "2":

        username = input("\nEnter your username: ")
        password = input("Create your password: ")

        user_numbers = []

        for uid in users:
            user_numbers.append(int(uid[1:]))

        if len(user_numbers) == 0:
            user_id = "U1"
        else:
            user_id = "U" + str(max(user_numbers) + 1)

        users[user_id] = {
            "username": username,
            "password": password
        }

        login_time = datetime.now()

        active_users[user_id] = {
            "username": username,
            "login_time": login_time
        }

        print("\nRegistration successful!")
        print("Your User ID is:", user_id)
        print("Login successful!")


    # =========================
    # 3. LOGOUT USER
    # =========================

    elif choice == "3":

        user_id = input("\nEnter User ID to logout: ")

        if user_id in active_users:

            username = active_users[user_id]["username"]
            login_time = active_users[user_id]["login_time"]

            logout_time = datetime.now()

            duration = logout_time - login_time

            duration_seconds = duration.total_seconds()
            duration_minutes = duration_seconds / 60

            session_data = pd.DataFrame([{
                "user_id": user_id,
                "username": username,
                "login_time": login_time,
                "logout_time": logout_time,
                "duration_seconds": round(duration_seconds, 2),
                "duration_minutes": round(duration_minutes, 2)
            }])

            try:
                old_data = pd.read_csv("user_sessions.csv")

                final_data = pd.concat(
                    [old_data, session_data],
                    ignore_index=True
                )

            except FileNotFoundError:
                final_data = session_data

            final_data.to_csv(
                "user_sessions.csv",
                index=False
            )

            del active_users[user_id]

            print("\nLogout successful!")

        else:
            print("\nUser is not currently logged in.")


    # =========================
    # 4. VIEW CSV
    # =========================

    elif choice == "4":

        try:
            final_data = pd.read_csv("user_sessions.csv")

            print("\nUser Sessions:")
            print(final_data)

        except FileNotFoundError:
            print("\nNo session data available.")


    # =========================
    # INVALID CHOICE
    # =========================

    else:
        print("\nInvalid choice. Please enter 1, 2, 3 or 4.")
```
