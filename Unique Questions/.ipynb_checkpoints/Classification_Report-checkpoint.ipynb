```python
import ast
while True:
    try:
        matrix = ast.literal_eval(input("Enter list: "))
        
        n = len(matrix)
        
        total = 0
        
        for i in matrix:
            for j in i:
                total += j
    
        for i in range(n):
            if len(matrix)==len(matrix[i]):
                tp = matrix[i][i]
                fp = 0
                for j in range(n):
         
                    if j != i:
                        fp += matrix[j][i]
            
                fn = 0
                for j in range(n):
                    if j != i:
                        fn += matrix[i][j]
                tn = total - (tp+fp+fn)
                if tp + fp == 0:
                    precision = 0
                else:
                    precision = tp / (tp + fp)
            
                if tp + fn == 0:
                    recall = 0
                else:
                    recall = tp / (tp + fn)
            
                if precision + recall == 0:
                    f1 = 0
                else:
                    f1 = 2 * precision * recall / (precision + recall)
            
                print(f"Value : {tp},  Precision : {round(precision, 2)},  Recall : {round(recall, 2)},  F1-Score : {round(f1, 2)}, False Positive Rate : {round((fp/(fp+tn))*100,2)}, False Negative Rate : {round((fn/(fn+tp))*100,2)}")
                print(f"tp : {tp}, fp : {fp}, fn : {fn}, tn : {tn}")
            else:
                print("Matrix is not of the required size")
                print("Invalid Matrix")
    except :
        print(f"Invalid input ")
```
