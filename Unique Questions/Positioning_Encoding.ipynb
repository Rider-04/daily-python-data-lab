```python
import math
length=int(input("Enter the length of vector : "))
pos=int(input("Enter the Word Position : "))
d=int(input("Enter the dimention : "))
v=[]
new_v=[]
for i in range(length):
    a=float(input(f"Enter the {i} index value : "))
    v.append(a)
    if i%2==0:
        even = math.sin(pos/(10000**((2*i)/d)))
        new_v.append(even)
    else :
        odd = math.cos(pos/(10000**(((2*i)+1)/d)))
        new_v.append(odd)
print(f"Old Vector : {v}")  
print(f"Positioning Encoded Vector : {new_v}")
```
