```python
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
```

```python
le = LabelEncoder()
```

```python
data["status_type"]=le.fit_transform(data["status_type"])
```

```python
sc = MinMaxScaler()
```

```python
data = pd.DataFrame(sc.fit_transform(data),columns=["status_type","num_reactions","num_comments","num_shares","num_likes","num_loves","num_wows","num_hahas","num_sads","num_angrys"])
```

```python
data
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans=KMeans(n_clusters=2,random_state=0)
```

```python
clust=kmeans.fit_transform(data)
```

```python
data["not_hyped"]= kmeans.labels_
```

```python
data
```

```python
kmeans.cluster_centers_
```

```python
kmeans.inertia_
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dc = DecisionTreeClassifier(max_depth=5)
```

```python
x= data.drop(columns=["not_hyped"])
```

```python
y=data["not_hyped"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size = 0.2, random_state = 42)
```

```python
dc.fit(X_train,y_train)
```

```python
ypred= dc.predict(X_test)
```

```python
ypred
```

```python
from sklearn.metrics import classification_report
```

```python
from xgboost import XGBClassifier
```

```python
XGB = XGBClassifier(max_depth =1,
                   max_leaves = 1,
                   n_estimators=50)
```

```python
XGB.fit(X_train,y_train)
```

```python
xgb_pred = XGB.predict(X_test)
```

```python
xgb_pred
```

```python
print(classification_report(y_test,xgb_pred))
```
